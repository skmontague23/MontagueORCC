```r
# read in merged data
Growth_Data_forR <- read_csv("MERGED_Summer_weights.csv", 
                             col_types = cols(Phase_1_treat = col_factor(),
                                              Phase_2_treat = col_factor(),
                                              Phase_2_rep_R = col_factor(),
                                              Phase_1_rep_R = col_factor()))
```

```r
mortality <- Growth_Data_forR%>%
mutate(Mortality = ifelse(is.na(Actual_tissue_growth_mg), 1, 0))

## 1 is dead, 0 is alive
```

```r
#visualize mortality

#format for jupyter
options(repr.plot.width = 12, repr.plot.height = 6)

summary_stats <- mortality %>%
  group_by(TreatPop) %>%
  mutate(
    mean_pre = mean(Mortality, na.rm = TRUE),
    se_pre = std.error(Mortality, na.rm = TRUE))

#plot TISSUE
ggplot(summary_stats, aes(x = TreatPop, y = mean_pre, color = TreatPop)) +
  geom_point(size = 4, position = position_dodge(0.9)) + # Plot means as points
  geom_errorbar(aes(ymin = mean_pre - se_pre, ymax = mean_pre + se_pre), 
                width = 0.1, position = position_dodge(0.9)) + # Error bars for SD
  theme_classic(base_size = 14)+
  labs(x = "Winter Treatment", y = "Proportion Spring Mortality") +
ylim(0,1)+
  theme(legend.position = "none") # Remove legend
```

```r
#glmer
m1 <- glmer(Mortality ~ Phase_1_treat * Phase_2_treat +
               (1 | Phase_1_rep_R) + (1 | Phase_2_rep_R),
             data = mortality,
             family = binomial)

aov_tab <- Anova(m1, type = "III", test = "Chisq")


#remove scientific notation
aov_tab %>%
  mutate(`Pr(>Chisq)` = round(`Pr(>Chisq)`, 4))
```
